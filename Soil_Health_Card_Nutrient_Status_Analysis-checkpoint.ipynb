# Soil Organic Carbon and Nutrient Status Analysis from Soil Health Card Data
**Real data:** national Soil Health Card (SHC) results, Ministry of Agriculture and Farmers Welfare, Government of India, published on the India Data Portal as *Soil Nutrient Analysis* (Open Data Commons Attribution License).

**What this data is.** It is *not* one row per soil sample with a measured number. It is village-level data in long format: each row gives, for one village, one nutrient and one rating (for example *Low*, *Medium*, *High*), how much of that village's test results fall in that rating (`value`). So this project analyses **soil fertility status** (what share of samples are rated Low, Medium or High) and cannot compute measured organic carbon percentages or carbon stock in tonnes per hectare.

**Workflow:** load one state in chunks -> inspect -> standardise -> remove duplicates -> validate -> check coverage (missing data) -> reshape -> organic carbon status by district and block -> hotspot villages -> other nutrients -> trends by year -> save outputs.

> **How to run:** put the downloaded CSV next to this notebook, set `FILE` and `STATE` in the first code cell, then run all cells top to bottom.

## 0. Settings (edit this cell only)

```python
import re, os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
try:
    from IPython.display import display
except ImportError:
    display = print

# ---------------- EDIT THESE ----------------
FILE = "soil_health_card.csv"   # name of the CSV you downloaded
STATE = "Telangana"             # state to analyse (as spelled in the data)
YEAR = "latest"                 # "latest", "all", or a specific value such as "2024-25"
CHUNK = 500_000                 # rows read at a time (lower it if you run out of memory)
VALUE_MODE = None               # None = detect automatically; or "counts" / "percent"
# --------------------------------------------

pd.set_option("display.max_columns", 40)
pd.set_option("display.width", 150)
sns.set_theme(style="whitegrid")
os.makedirs("figures", exist_ok=True)

log = []
def add_log(step, column, rows, action):
    log.append({"Step": step, "Column": column, "Rows_affected": int(rows), "Action": action})
```

## 1. Look at the file

```python
NEEDED = ["year", "state_name", "district_name", "block_name", "village_code", "village_name",
          "nutrient_name", "nutrient_level", "value"]
head = pd.read_csv(FILE, nrows=5)
print("Columns:", list(head.columns))
absent = [c for c in NEEDED if c not in head.columns]
assert not absent, f"These expected columns are missing from the file: {absent}"
head
```

## 2. Load one state in chunks
The full file is about 1.2 GB, so each chunk is filtered to the chosen state and the rest is dropped.

```python
parts, total_rows, all_states = [], 0, set()
for chunk in pd.read_csv(FILE, usecols=NEEDED, chunksize=CHUNK, low_memory=False):
    total_rows += len(chunk)
    all_states.update(chunk["state_name"].dropna().astype(str).str.strip().unique())
    keep = chunk["state_name"].astype(str).str.strip().str.lower() == STATE.lower()
    if keep.any():
        parts.append(chunk[keep])

if not parts:
    print("No rows matched STATE. States found in the file:")
    print(sorted(all_states))
    raise SystemExit("Set STATE to one of the names above.")

raw = pd.concat(parts, ignore_index=True)
print(f"Rows in file: {total_rows:,} | rows for {STATE}: {len(raw):,}")
add_log("Load", "all", len(raw), f"Rows kept for {STATE} (file had {total_rows:,} rows)")
df = raw.rename(columns={"state_name": "state", "district_name": "district", "block_name": "block",
                         "nutrient_name": "nutrient_raw", "nutrient_level": "level_raw"}).copy()
df.head()
```

## 3. Standardise names and ratings
Place names are trimmed and given one capitalisation so the same place is not split into several groups. Nutrient names are mapped to short standard names, and the ratings *Low / Medium / High* are written the same way everywhere. Other ratings (for example *Acidic* or *Sufficient*) are kept as they are. **Read the table printed at the end:** it shows every nutrient and rating found in your data.

```python
for c in ["state", "district", "block", "village_name"]:
    before = df[c].astype(str)
    df[c] = df[c].astype(str).str.strip().str.title().replace({"Nan": np.nan, "": np.nan})
    add_log("Standardise", c, (before != df[c].fillna("")).sum(), "Trimmed spaces and unified capitalisation")

def std_nutrient(name):
    t = re.sub(r"[^a-z0-9 ]", " ", str(name).lower()).strip()
    if "organic" in t or t in ("oc", "o c"):     return "OC"
    if "nitrogen" in t:                          return "N"
    if "phosph" in t or "p2o5" in t:             return "P"
    if "potass" in t or "k2o" in t:              return "K"
    if "sulph" in t or "sulf" in t:              return "S"
    if "zinc" in t:                              return "Zn"
    if "boron" in t:                             return "B"
    if "iron" in t:                              return "Fe"
    if "copper" in t:                            return "Cu"
    if "mangan" in t:                            return "Mn"
    if "conductiv" in t or t in ("ec", "salinity"): return "EC"
    if re.fullmatch(r"(soil )?ph", t) or "soil ph" in t: return "pH"
    return str(name).strip().title()

def std_level(x):
    t = str(x).strip().lower()
    if t in ("low", "l"):                  return "Low"
    if t in ("medium", "med", "moderate"): return "Medium"
    if t in ("high", "h"):                 return "High"
    return str(x).strip().title()

df["nutrient"] = df["nutrient_raw"].map(std_nutrient)
df["level"] = df["level_raw"].map(std_level)
df["value"] = pd.to_numeric(df["value"], errors="coerce")
add_log("Convert", "value", df["value"].isna().sum(), "Non-numeric values turned into blanks")

print("Nutrients and ratings found (number of rows):")
display(pd.crosstab(df["nutrient"], df["level"]))
print("Original nutrient names:", sorted(df["nutrient_raw"].astype(str).unique()))
print("Years:", sorted(df["year"].astype(str).unique()))
```

## 4. Remove duplicates
The same village, year, nutrient and rating should appear once. Exact repeats are dropped (first kept). If the same combination appears with *different* values, that is a real conflict: it is counted and reported, and the first copy is kept.

```python
key = ["year", "village_code", "nutrient_raw", "level_raw"]
dup_all = df.duplicated(subset=key, keep=False)
conflicts = int(df[dup_all].groupby(key)["value"].nunique(dropna=False).gt(1).sum())
before = len(df)
df = df.drop_duplicates(subset=key, keep="first").reset_index(drop=True)
removed = before - len(df)
add_log("Duplicates", "village+year+nutrient+rating", removed, "Repeated rows removed (first copy kept)")
print(f"Duplicate rows removed: {removed:,} | key combinations with conflicting values: {conflicts}")
```

## 5. Validate values
A share or a count cannot be negative, and blanks cannot be used. Negative values are set to blank and flagged, then rows without a value are dropped from the calculations (they are never replaced by guesses). The notebook also works out whether `value` holds **sample counts** or **percentages**, because that decides how districts are summarised: counts can be added up exactly, percentages can only be averaged.

```python
neg = df["value"] < 0
add_log("Validate", "value", neg.sum(), "Negative values set to blank")
df.loc[neg, "value"] = np.nan
n_blank = int(df["value"].isna().sum())
add_log("Missing", "value", n_blank, "Rows without a usable value excluded from calculations (not filled)")
df = df.dropna(subset=["value", "village_code", "district"]).reset_index(drop=True)
print(f"Negative values blanked: {int(neg.sum())} | rows without value or place removed: {n_blank}")

sums = df.groupby(["year", "village_code", "nutrient"])["value"].sum()
share_100 = sums.between(99, 101).mean()
detected = "percent" if share_100 > 0.9 else "counts"
MODE = VALUE_MODE or detected
print(f"Share of village-nutrient groups that add up to about 100: {share_100:.0%}")
print(f"value is treated as: {MODE}   (detected: {detected})")
print("value summary:"); print(df["value"].describe().round(2))
```

## 6. Coverage: which villages have which nutrient?
A village that has no organic carbon result is a **gap**, not a Low rating. Gaps are counted here and left blank; they never enter the averages.

```python
n_villages = df["village_code"].nunique()
cov = (df.groupby("nutrient")["village_code"].nunique() / n_villages * 100).round(1).sort_values()
cov = cov.rename("Villages_with_data_%").to_frame()
print(f"Villages in {STATE}: {n_villages:,}")
cov
```

## 7. Choose the year and reshape

```python
years = sorted(df["year"].astype(str).unique())
if YEAR == "latest":   an_year = [years[-1]]
elif YEAR == "all":    an_year = years
else:                  an_year = [str(YEAR)]
an = df[df["year"].astype(str).isin(an_year)].copy()
print("Years used for the main analysis:", an_year, "| rows:", f"{len(an):,}")

# village-level percentage in each rating
grp = ["year", "village_code", "nutrient"]
an["total"] = an.groupby(grp)["value"].transform("sum")
if MODE == "counts":
    an["pct"] = np.where(an["total"] > 0, an["value"] / an["total"] * 100, np.nan)
else:
    an["pct"] = an["value"]

def level_shares(data, by):
    """Share of samples (%) in each rating, per group and nutrient."""
    if MODE == "counts":
        s = data.groupby(by + ["nutrient", "level"])["value"].sum().unstack("level")
        n = s.sum(axis=1).rename("Samples")
        share = s.div(s.sum(axis=1), axis=0) * 100
        share = share.join(n)
    else:
        share = data.groupby(by + ["nutrient", "level"])["pct"].mean().unstack("level")
    share["Villages"] = data.groupby(by + ["nutrient"])["village_code"].nunique()
    return share.round(2)

def nutrient_index(share):
    """Nutrient Index Value = (1*Low + 2*Medium + 3*High) / 100. <1.67 low, 1.67-2.33 medium, >2.33 high."""
    cols = ["Low", "Medium", "High"]
    if not all(c in share.columns for c in cols):
        return pd.Series(np.nan, index=share.index)
    s = share[cols].fillna(0)
    return ((s["Low"] * 1 + s["Medium"] * 2 + s["High"] * 3) / 100).round(2)
print("Reshaping done.")
```

## 8. Overall nutrient status for the state

```python
state_share = level_shares(an, ["state"]).reset_index().drop(columns="state").set_index("nutrient")
state_share["Nutrient_index"] = nutrient_index(state_share)
state_share.to_csv("nutrient_status.csv")
state_share
```

```python
lmh = [c for c in ["Low", "Medium", "High"] if c in state_share.columns]
sub = state_share[state_share[lmh].notna().all(axis=1)] if lmh else state_share.iloc[0:0]
sub = sub[sub[lmh].sum(axis=1) > 50]
if len(sub):
    sub = sub.sort_values("Low")
    sub[lmh].plot(kind="barh", stacked=True, figsize=(9, 0.5 * len(sub) + 2),
                  color=["#C00000", "#FFC000", "#548235"][:len(lmh)])
    plt.xlabel("Share of test results (%)"); plt.ylabel("")
    plt.title(f"Nutrients rated Low / Medium / High, {STATE} ({', '.join(an_year)})")
    plt.legend(title="Rating", bbox_to_anchor=(1.02, 1), loc="upper left")
    plt.tight_layout(); plt.savefig("figures/nutrient_status.png", dpi=150); plt.show()
else:
    print("No nutrient with Low/Medium/High ratings found; check the table in section 3.")
```

## 9. Soil organic carbon (SOC) status
Organic carbon is rated **Low / Medium / High** in the Soil Health Card system. This section shows, for each district, what share of samples are rated Low, Medium or High, and a **Nutrient Index Value** `(1 x Low% + 2 x Medium% + 3 x High%) / 100`, a standard way to summarise ratings in one number (below 1.67 = low fertility, 1.67 to 2.33 = medium, above 2.33 = high). Villages with no organic carbon result are excluded, not counted as Low.

```python
oc = an[an["nutrient"] == "OC"].copy()
if oc.empty:
    print("No organic carbon rows found. Nutrient names in your data:")
    print(sorted(df["nutrient_raw"].astype(str).unique()))
    raise SystemExit("Organic carbon not found: see the list above.")
print("Organic carbon ratings found:", sorted(oc["level"].unique()))
print(f"Villages with an organic carbon result: {oc['village_code'].nunique():,} of {n_villages:,}")

dist = level_shares(oc, ["district"]).reset_index().drop(columns="nutrient").set_index("district")
dist["Nutrient_index"] = nutrient_index(dist)
min_n = 30 if MODE == "counts" else 5
size_col = "Samples" if MODE == "counts" else "Villages"
dist_ok = dist[dist[size_col] >= min_n].sort_values("Low", ascending=False)
dist_ok.to_csv("oc_status_by_district.csv")
print(f"{len(dist_ok)} districts with enough data ({size_col} >= {min_n}); sorted from highest Low share")
dist_ok
```

```python
cols = [c for c in ["Low", "Medium", "High"] if c in dist_ok.columns]
plot_d = dist_ok[cols].fillna(0)
if len(plot_d) > 30:
    plot_d = pd.concat([plot_d.head(15), plot_d.tail(15)])
plot_d = plot_d.iloc[::-1]
plot_d.plot(kind="barh", stacked=True, figsize=(9, max(4, 0.32 * len(plot_d))),
            color=["#C00000", "#FFC000", "#548235"][:len(cols)])
plt.xlabel("Share of organic carbon results (%)"); plt.ylabel("")
plt.title(f"Organic carbon rating by district, {STATE} ({', '.join(an_year)})")
plt.legend(title="Rating", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout(); plt.savefig("figures/oc_by_district.png", dpi=150); plt.show()
```

```python
# Blocks inside the district with the highest Low share
worst = dist_ok.index[0]
blk = level_shares(oc[oc["district"] == worst], ["block"]).reset_index().drop(columns="nutrient").set_index("block")
blk["Nutrient_index"] = nutrient_index(blk)
blk = blk.sort_values("Low", ascending=False)
blk.to_csv("oc_status_by_block_worst_district.csv")
print(f"Blocks in {worst} (the district with the highest Low organic carbon share):")
blk
```

## 10. Hotspot villages
Villages where most organic carbon results are rated Low are the first places to target for compost, green manure or residue management. Only villages with enough results are listed, so a village with two samples does not dominate the list.

```python
low_oc = oc[oc["level"] == "Low"][["village_code", "village_name", "district", "block", "pct", "total"]].copy()
low_oc = low_oc.rename(columns={"pct": "Low_OC_pct", "total": "Results_in_village"})
min_village = 10 if MODE == "counts" else 0
hot = low_oc[(low_oc["Low_OC_pct"] >= 50) & (low_oc["Results_in_village"] >= min_village)]
hot = hot.sort_values(["Low_OC_pct", "Results_in_village"], ascending=False)
hot.head(200).to_csv("hotspot_villages_low_oc.csv", index=False)
n_hot, n_all = hot["village_code"].nunique(), oc["village_code"].nunique()
print(f"Villages where 50% or more of results are Low: {n_hot:,} of {n_all:,} ({n_hot / max(n_all, 1) * 100:.1f}%)")
hot.head(15)
```

```python
v = oc[oc["level"] == "Low"]["pct"].dropna()
plt.figure(figsize=(7, 4))
sns.histplot(v, bins=20, color="#C00000")
plt.xlabel("Share of results rated Low organic carbon in a village (%)"); plt.ylabel("Number of villages")
plt.title("How widespread is Low organic carbon across villages?")
plt.tight_layout(); plt.savefig("figures/village_low_oc_hist.png", dpi=150); plt.show()
```

## 11. Is Low organic carbon linked to other nutrient problems?
For each village, the share of results rated Low is compared across nutrients. A positive correlation means villages that are low in one nutrient tend to be low in the other. It shows association, not cause.

```python
low_tbl = (an[an["level"] == "Low"].pivot_table(index="village_code", columns="nutrient", values="pct", aggfunc="first"))
low_tbl = low_tbl.dropna(axis=1, thresh=int(0.5 * len(low_tbl)))
if low_tbl.shape[1] >= 2:
    corr = low_tbl.corr()
    plt.figure(figsize=(7, 6))
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1, square=True)
    plt.title("Correlation of village-level Low share between nutrients")
    plt.tight_layout(); plt.savefig("figures/low_share_correlation.png", dpi=150); plt.show()
    pairs = corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1)).stack().sort_values(key=abs, ascending=False)
    print("Strongest relationships:"); print(pairs.head(5).round(2))
else:
    print("Not enough nutrients with Low ratings to compare.")
```

## 12. Soil reaction (pH) and salinity (EC)

```python
for nut, label in [("pH", "Soil pH classes"), ("EC", "Salinity (EC) classes")]:
    s = an[an["nutrient"] == nut]
    if s.empty:
        print(label, ": not found in the data"); continue
    sh = level_shares(s, ["state"]).reset_index().drop(columns=["state", "nutrient", "Samples", "Villages"], errors="ignore")
    print(label + " (share of results, %)"); display(sh.T.rename(columns={0: "Share_%"}))
```

## 13. Change over time
If the data holds more than one year, the state-wide share of Low organic carbon is compared across years. A change can reflect more or different villages being tested, so treat it as a hint and check the number of results each year.

```python
oc_all = df[df["nutrient"] == "OC"].copy()
oc_all["total"] = oc_all.groupby(["year", "village_code", "nutrient"])["value"].transform("sum")
oc_all["pct"] = np.where(oc_all["total"] > 0, oc_all["value"] / oc_all["total"] * 100, np.nan) if MODE == "counts" else oc_all["value"]
trend = level_shares(oc_all, ["year"]).reset_index().drop(columns="nutrient").set_index("year")
trend.to_csv("oc_trend_by_year.csv")
if len(trend) > 1 and "Low" in trend.columns:
    trend[[c for c in ["Low", "Medium", "High"] if c in trend.columns]].plot(kind="bar", figsize=(7, 4),
        color=["#C00000", "#FFC000", "#548235"][:len([c for c in ["Low", "Medium", "High"] if c in trend.columns])])
    plt.ylabel("Share of results (%)"); plt.title("Organic carbon rating by year")
    plt.tight_layout(); plt.savefig("figures/oc_trend.png", dpi=150); plt.show()
else:
    print("Only one year in the data, so no trend can be shown.")
trend
```

## Limitations
- The data holds **ratings (Low / Medium / High), not measured values**, so organic carbon percentages and carbon stock cannot be calculated. The analysis describes fertility *status*.
- Ratings depend on the cut-offs used by the testing labs. This project does not recompute them.
- Results come from many labs and collectors; differences between districts can reflect testing practice as well as real soil differences.
- Villages that were not tested are absent, not Low. Coverage is shown in section 6.
- When `value` is a percentage (not a count), district figures are simple averages across villages, so large and small villages count equally.
- The Nutrient Index Value cut-offs (1.67 and 2.33) are the commonly used ones; check them against your guideline.
- Correlations describe association only, not cause.
